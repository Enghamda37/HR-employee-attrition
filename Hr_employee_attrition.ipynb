```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries loaded successfully!")
```

```python

df = pd.read_csv("WA_Fn-UseC_-HR-Employee-Attrition.csv")

print("Dataset shape:", df.shape)

display(df.head())
```

```python

print("Columns:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())
```

```python
df.isnull().sum()
```

```python

print("Missing values:")
print(df.isnull().sum().sum())

print("Duplicate rows:", df.duplicated().sum())
```

```python
#tirada shqaalo ee tahay io kuwa jooga
print(df["Attrition"].value_counts())

sns.countplot(data=df, x="Attrition")
plt.title("Employee Attrition Analysis")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")
plt.show()
```

```python
#tirada shaqaalaha
print("Total employees:", len(df))

print("\nEmployees by department:")
print(df["Department"].value_counts())

print("\nAttrition distribution:")
print(df["Attrition"].value_counts())
```

```python
#employee attrition bt departmnets

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Department",
    hue="Attrition"
)

plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python
#attrition rate
attrition_rate = (
    df["Attrition"].eq("Yes").mean() * 100
)

print(f"Employee Attrition Rate: {attrition_rate:.2f}%")
```

```python
#Monthly Income Analysis
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Attrition",
    y="MonthlyIncome"
)

plt.title("Monthly Income by Employee Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")
plt.show()
```

```python

Q1 = df["MonthlyIncome"].quantile(0.25)
Q3 = df["MonthlyIncome"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["MonthlyIncome"] < lower_bound) |
    (df["MonthlyIncome"] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of outliers:", len(outliers))

display(outliers[["MonthlyIncome", "Department", "JobRole"]].head(10))
```

```python

# Keep the original dataset unchanged
df_original = df.copy()

# Create a separate dataset without MonthlyIncome outliers
df_no_outliers = df[
    (df["MonthlyIncome"] >= lower_bound) &
    (df["MonthlyIncome"] <= upper_bound)
].copy()

print("Original rows:", len(df_original))
print("Outliers identified:", len(outliers))
print("Rows after removing outliers:", len(df_no_outliers))
```

```python

# Check the highest monthly incomes
display(
    outliers[
        ["MonthlyIncome", "Department", "JobRole", "JobLevel"]
    ].sort_values(
        by="MonthlyIncome",
        ascending=False
    ).head(15)
)
```

```python

attrition_by_department = pd.crosstab(
    df["Department"],
    df["Attrition"]
)

attrition_by_department["Attrition_Rate_%"] = (
    attrition_by_department["Yes"] /
    attrition_by_department.sum(axis=1).where(
        attrition_by_department.sum(axis=1) > 0
    ) * 100
)

display(attrition_by_department)

attrition_by_department["Attrition_Rate_%"].plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

```python

rates = attrition_by_department["Attrition_Rate_%"].sort_values(
    ascending=False
)

ax = rates.plot(kind="bar", figsize=(9, 5))

plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)

for bar in ax.patches:
    ax.annotate(
        f"{bar.get_height():.1f}%",
        (bar.get_x() + bar.get_width() / 2,
         bar.get_height()),
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python

overtime_analysis = pd.crosstab(
    df["OverTime"],
    df["Attrition"],
    normalize="index"
) * 100

print(overtime_analysis.round(2))

overtime_analysis.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Attrition by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Percentage of Employees (%)")
plt.xticks(rotation=0)
plt.legend(title="Attrition")
plt.tight_layout()
plt.show()
```

```python

# 1. Overall attrition
attrition_rate = (df["Attrition"] == "Yes").mean() * 100
print(f"Overall attrition rate: {attrition_rate:.2f}%")

# 2. Attrition by department
dept_rate = (
    df.assign(Left=df["Attrition"].eq("Yes"))
      .groupby("Department")["Left"]
      .mean()
      .mul(100)
      .sort_values(ascending=False)
)

print("\nAttrition rate by department:")
print(dept_rate.round(2))

# 3. Average monthly income by attrition
income_analysis = df.groupby("Attrition")["MonthlyIncome"].mean()
print("\nAverage monthly income by attrition:")
print(income_analysis.round(2))

# 4. Attrition by overtime
overtime_rate = (
    df.assign(Left=df["Attrition"].eq("Yes"))
      .groupby("OverTime")["Left"]
      .mean()
      .mul(100)
)

print("\nAttrition rate by overtime:")
print(overtime_rate.round(2))
```
